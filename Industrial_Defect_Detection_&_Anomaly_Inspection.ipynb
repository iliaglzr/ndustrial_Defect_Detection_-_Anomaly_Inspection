```python

from pathlib import Path
from collections import Counter

DATASET = ROOT / "datasets" / "neu_yolo"

SPLITS = ["train", "validation"]
NUM_CLASSES = 6

for split in SPLITS:
    image_dir = DATASET / "images" / split
    label_dir = DATASET / "labels" / split

    print(f"\n========== {split.upper()} ==========")

    if not image_dir.exists() or not label_dir.exists():
        print("ERROR: Images or labels folder is missing.")
        continue

    images = [
        p for p in image_dir.iterdir()
        if p.is_file()
        and p.suffix.lower() in [".jpg", ".jpeg", ".png", ".bmp"]
    ]

    labels = list(label_dir.glob("*.txt"))

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in labels}

    missing_labels = image_stems - label_stems
    missing_images = label_stems - image_stems

    print("Images:", len(images))
    print("Labels:", len(labels))
    print("Images without labels:", len(missing_labels))
    print("Labels without images:", len(missing_images))

    class_counts = Counter()
    empty_labels = 0
    invalid_lines = 0
    total_boxes = 0

    for label_file in labels:
        content = label_file.read_text(
            encoding="utf-8"
        ).strip()

        if not content:
            empty_labels += 1
            continue

        for line in content.splitlines():
            parts = line.split()

            if len(parts) != 5:
                invalid_lines += 1
                continue

            try:
                class_id = int(parts[0])
                x, y, w, h = map(float, parts[1:])
            except ValueError:
                invalid_lines += 1
                continue

            if not 0 <= class_id < NUM_CLASSES:
                invalid_lines += 1
                continue

            if not all(0 <= v <= 1 for v in [x, y, w, h]):
                invalid_lines += 1
                continue

            if w <= 0 or h <= 0:
                invalid_lines += 1
                continue

            class_counts[class_id] += 1
            total_boxes += 1

    print("Total bounding boxes:", total_boxes)
    print("Empty labels:", empty_labels)
    print("Invalid annotation lines:", invalid_lines)

    print("\nObjects per class:")
    for class_id in range(NUM_CLASSES):
        print(f"Class {class_id}: {class_counts[class_id]}")

    if missing_labels or missing_images or invalid_lines:
        print("\nRESULT: Needs investigation.")
    else:
        print("\nRESULT: Basic checks passed.")
```

```python
from pathlib import Path

# مسیر دیتاست تبدیل‌شده در Google Drive
DATASET = Path("/content/drive/MyDrive/NEU-DET/datasets/neu_yolo")
YAML_PATH = DATASET / "neu.yaml"

# بررسی وجود پوشه‌ها
for folder in [
    DATASET / "images" / "train",
    DATASET / "images" / "validation",
    DATASET / "labels" / "train",
    DATASET / "labels" / "validation",
]:
    print(f"{folder}: {'OK' if folder.is_dir() else 'NOT FOUND'}")

# ساخت فایل تنظیمات با مسیر صحیح Colab
yaml_content = f"""path: {DATASET}
train: images/train
val: images/validation

names:
  0: crazing
  1: inclusion
  2: patches
  3: pitted_surface
  4: rolled-in_scale
  5: scratches
"""

YAML_PATH.write_text(yaml_content, encoding="utf-8")

print("\nUpdated YAML:")
print(YAML_PATH.read_text(encoding="utf-8"))

# بررسی اینکه تصاویر واقعاً وجود دارند
train_images = list((DATASET / "images" / "train").glob("*"))
val_images = list((DATASET / "images" / "validation").glob("*"))

print("\nTrain images:", len(train_images))
print("Validation images:", len(val_images))

assert len(train_images) > 0, "Training images were not found!"
assert len(val_images) > 0, "Validation images were not found!"

print("\nDataset paths look good!")
```

```python
from ultralytics import YOLO
import torch
from pathlib import Path

# مسیر فایل تنظیمات دیتاست
DATA_YAML = "/content/drive/MyDrive/NEU-DET/datasets/neu_yolo/neu.yaml"

# بررسی GPU
if torch.cuda.is_available():
    device = 0
    print("GPU:", torch.cuda.get_device_name(0))
else:
    device = "cpu"
    print("GPU در دسترس نیست؛ آموزش با CPU انجام می‌شود.")

# بارگذاری مدل از وزن‌های ازپیش‌آموزش‌دیده
model = YOLO("yolo26n.pt")

# شروع آموزش
results = model.train(
    data=DATA_YAML,
    epochs=30,
    imgsz=640,
    batch=16,
    device=device,
    workers=2,
    patience=10,
    project="/content/drive/MyDrive/NEU-DET/runs",
    name="neu_defect_detection",
    plots=True,
    save=True
)
```
